# Run Plan demo (R)

This notebook exists to try the Run Plan button. Each code cell prints a counter and its own label, so the order in which the cells ran can be read from the outputs. Nothing is written to disk.

The cells are deliberately out of order: `save-checkpoint` sits above the cells whose result it stores. Three cells are named with a `#| label:` comment and two with a `label:` cell tag.

```r
#| label: setup
demo_step <- 0
tick <- function(label) {
  demo_step <<- demo_step + 1
  cat(sprintf("[%d] %s\n", demo_step, label))
}
tick("setup")
```

```r
tick("save-checkpoint")
checkpoint <- list(data = d, total = total)
cat("checkpoint holds", nrow(checkpoint$data), "rows and total", checkpoint$total, "\n")
```

```r
#| label: summarise
tick("summarise")
total <- sum(d$value)
cat("total is", total, "\n")
```

```r
tick("load-data")
d <- data.frame(id = 1:3, value = c(10, 20, 30))
```
